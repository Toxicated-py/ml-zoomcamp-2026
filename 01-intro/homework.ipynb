Q6: Horsepower median before/after filling missing values

```python
median_before = df['horsepower'].median
median_before
```

```python
most_frequent = df['horsepower'].mode()[0]
most_frequent
```

```python
df_q6['horsepower'] = df_q6['horsepower'].fillna(most_frequent)
```

```python
median_after = df_q6['horsepower'].median()
median_after
```
